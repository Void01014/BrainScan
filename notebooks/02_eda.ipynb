```python
import sys
sys.path.append("..")
from collections import Counter

import matplotlib.pyplot as plt
import numpy as np

from src.data_loader import load_dataset

images, labels, classes = load_dataset("../data/<same path as 01_loading>")

counts = Counter(labels)
print(counts)
print(f"Imbalance ratio (max/min): {max(counts.values()) / min(counts.values()):.2f}")

plt.figure(figsize=(6, 4))
plt.bar(counts.keys(), counts.values())
plt.title("Images per class")
plt.ylabel("Count")
plt.savefig("../docs/class_distribution.png", bbox_inches="tight")
plt.show()

fig, axes = plt.subplots(len(classes), 5, figsize=(12, 2.5 * len(classes)))
for i, cls in enumerate(classes):
    idx = np.where(labels == cls)[0][:5]
    for j, k in enumerate(idx):
        axes[i, j].imshow(images[k])
        axes[i, j].axis("off")
    axes[i, 0].set_title(cls, loc="left")
plt.savefig("../docs/samples_per_class.png", bbox_inches="tight")
plt.show()
```
